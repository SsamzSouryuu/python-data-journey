```python
# Histogram, Sebaran satu kolom angka

import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv('hasil_data_lab2.csv')

plt.hist(data['kadar_air'], bins=10)
plt.xlabel('Kadar Air (%)')
plt.ylabel('Jumlah Sampel')
plt.title('Sebaran Kadar Air')
plt.show()
```

Ada 1 sampel di antara 41-43, itu outlier di modul 9 kemaren, yaitu keripik apel yang angkanya ekstream. Dan benar hasil perhitungan, kadar airnya 40.21739130434783. Lalu ada 8 sampel yang kadar airnya 53-55, dst.

```python
# Box Plot, Konsep IQR, Q1-Q3. Median, Outlier, tanpa code manual dan rumus manual.

import seaborn as sns
import pandas as pd

data = pd.read_csv('hasil_data_lab2.csv')

sns.boxplot(y=data['kadar_air'])
plt.title('Boxplot Kadar Air')
plt.show()
```

Kotak biru → itu area Q1-Q3 (dari ~55 sampai ~62.5), alias 50% data yang paling "biasa"

Garis di tengah kotak (sekitar 59.5) → itu median

Garis tipis di atas & bawah kotak (disebut "whisker") → itu batas wajar data (mirip konsep batas_atas/batas_bawah)

Lingkaran kecil sendirian di bawah (~40) → outlier, otomatis ke-plot karena di luar whisker

```python
# Bar Chart, lihat jumlah grade dari sampel hasil lab.

import pandas as pd

data = pd.read_csv('hasil_data_lab2.csv')

data['grade'].value_counts().plot(kind='bar')
plt.xlabel('Grade')
plt.ylabel('Jumlah Sampel')
plt.title('Jumlah Sampel per Grade')
plt.show()
```

```python
# Bar Chart, menentukan rata rata selamat per kelas kapal

import pandas as pd
import matplotlib.pyplot as plt

data_titanic = pd.read_csv('kapal_titanic.csv')

rata_per_kelas = data_titanic.groupby('pclass')['survived'].mean()
rata_per_kelas.plot(kind='bar')
plt.xlabel('Kelas')
plt.ylabel('Tingkat Selamat')
plt.title('Tingkat Selamat per Kelas Kapal')
plt.show()
```

```python
# Scatter Plot, Hubungan antara 2 angka/2 variabel.

import pandas as pd
import matplotlib.pyplot as plt

data_titanic = pd.read_csv('kapal_titanic.csv')

plt.scatter(data_titanic['fare'], data_titanic['survived'])
plt.xlabel('Harga Tiket')
plt.ylabel('Selamat (1) / Tidak (0)')
plt.title('Harga Tiket vs Peluang Selamat')
plt.show()
```
