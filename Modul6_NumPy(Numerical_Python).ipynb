Nah, NumPy ini ibarat mesin scanner raksasa. Lempar 10 juta data sekaligus ke dalam wadahnya (Array), dan mesin bisa ngitung rata-rata atau standar deviasi semuanya sekaligus. Nggak butuh looping.

```python
import numpy as np

# 1. Masukin data kadar air sampel kita ke dalam wadah Array NumPy
kadar_air_sampel = np.array([10, 12, 14, 16, 18])

# 2. Cari rata-ratanya cukup pakai satu jurus ini!
rata_rata = kadar_air_sampel.mean()

print(f"Hasil Kadar Air: {kadar_air_sampel}")
print(f"Rata-rata (Mean) otomatis: {rata_rata}")
```

Di NumPy, wadah khusus penyimpan deretan angka itu namanya "array". Begitu data udah masuk ke dalam array NumPy, nyari rata-rata tinggal manggil satu fungsi bawaan yang namanya .mean().

```python
import numpy as np

# Data array
kadar_air_sampel = np.array([10, 12, 14, 16, 18])

# 1. Cari Rata-rata
rata_rata = kadar_air_sampel.mean()

# 2. Cari Standar Deviasi (Ganti ??? dengan fungsinya)
standar_deviasi = kadar_air_sampel.std()

print(f"Rata-rata: {rata_rata}")
print(f"Standar Deviasi: {standar_deviasi}")
```

Selain rata-rata, biasanya ada juga: "Seberapa jauh sih data kadar air di tiap sampel itu mencar atau bervariasi dari nilai rata-ratanya?"

Di statistik, ukuran penyebaran data ini dinamakan Standar Deviasi (Standard Deviation).

Di dunia pabrik (terutama Quality Control), konsistensi segalanya. Kalau standar deviasinya kecil, artinya semua keripik kadar airnya stabil di kisaran 14. Kalau standar deviasinya gede, artinya ada keripik yang terlalu basah (jamuran) dan ada yang terlalu kering (gosong). Makin nyebar, makin hancur kualitasnya.

```python
# Percobaan kedua (agar angka deviasi tidak panjang)

import numpy as np

# Data array
kadar_air_sampel = np.array([10, 12, 14, 16, 18])

# 1. Cari Rata-rata
rata_rata = kadar_air_sampel.mean()

# 2. Cari Standar Deviasi (Ganti ??? dengan fungsinya)
standar_deviasi = kadar_air_sampel.std()

print(f"Rata-rata: {rata_rata}")
print(f"Standar Deviasi: {standar_deviasi: .3f}")
```

Apa sih arti angka 2.82 ini?
Artinya, rata-rata simpangan (kemenceng-an) kadar air keripik lo dari nilai idealnya (14) adalah sekitar 2.8.

Berarti sebagian besar keripik lo kadar airnya ada di range 11.18 sampai 16.82 (didapet dari 14 dikurang 2.8, dan 14 ditambah 2.8).

"Standar deviasinya harus ditekan di bawah 1.0" berarti harus nyuruh orang lapangan buat ngegoreng keripiknya lebih seragam lagi biar kualitasnya konsisten.

```python
# Kapal Titanic (pandas)

# 1. Panggil library pandas
import pandas as pd

# 2. Baca file csv-nya dan simpan ke dalam kotak kardus bernama 'tabel_titanic'
tabel_titanic = pd.read_csv('kapal_titanic.csv')

# 3. Intip 5 baris teratas dari tabelnya
print(tabel_titanic.head())
```

```python
# 1. Bikin filter: ambil tabelnya, lalu saring kolom 'survived' yang nilainya ???
penumpang_selamat = tabel_titanic[tabel_titanic['survived'] == 1]

# 2. Kita hitung ada berapa baris (berapa orang) sih yang selamat?
jumlah_selamat = len(penumpang_selamat)
print(f"Total penumpang yang selamat ada: {jumlah_selamat} orang")

# 3. Intip 5 data teratas penumpang yang selamat
print("\nData Penumpang Selamat:")
print(penumpang_selamat.head())
```

```python
# 1. Comot HANYA kolom umur dari tabel penumpang yang selamat
umur_selamat = penumpang_selamat['age']

# 2. Hitung rata-ratanya pakai fungsi dari Modul 6!
rata_umur = umur_selamat.mean()

print(f"Rata-rata umur penumpang yang selamat: {rata_umur: .2f} tahun")
```

```python
# 1. Comot HANYA kolom jenis kelamin dari tabel penumpang yang selamat
gender_selamat = penumpang_selamat['sex']

# 2. Hitung jumlah masing-masing gender pakai fungsi sakti Pandas!
jumlah_per_gender = gender_selamat.value_counts()

print("Rincian gender penumpang selamat:")
print(jumlah_per_gender)
```

```python
# Kelompokkan penumpang selamat berdasarkan 'Kelas Tiket', comot 'Umur', lalu hitung rata-ratanya
rata_umur_per_kelas = penumpang_selamat.groupby('pclass')['age'].mean()

print("Rata-rata umur penumpang selamat berdasarkan Kelas Tiket (pclass):")
print(rata_umur_per_kelas)
```

```python
import numpy as np

kadar_air = np.array([15, 20, 18, 23, 21, 42, 12, 23, 20, 21, 32, 12 ,12, 15, 17])

rata_rata = kadar_air.mean()

standard_deviasi = kadar_air.std()

Batas_Atas_UCL = rata_rata + (2 * standard_deviasi)
Batas_Bawah_LCL = rata_rata - (2 * standard_deviasi)

print(f"Rata-rata: {rata_rata}")
print(f"Standar Deviasi: {standard_deviasi: .3f}")
print(f"Batas Atas UCL: {round(Batas_Atas_UCL)}")
print(f"Batas Bawah LCL: {Batas_Bawah_LCL: .3f}")

print(kadar_air > Batas_Atas_UCL)
print(kadar_air < Batas_Bawah_LCL)

#Nentuin/nyari angka berapa di array yang diatas ucl
sampel_diatas_ucl = kadar_air[kadar_air > Batas_Atas_UCL]
print(f"Sampel diatas UCL: {sampel_diatas_ucl}")

#Nyari angka berapa di array yang dibawah lcl
sampel_dibawah_lcl = kadar_air[kadar_air < Batas_Bawah_LCL]
print(f"Sampel dibawah LCL: {sampel_dibawah_lcl}")
```
