```python
# Kurva Lonceng

import numpy as np
import matplotlib.pyplot as plt

data_simulasi = np.random.normal(loc=20, scale=5, size=1000)  # simulasi 1000 data, mean=20, std=5
plt.hist(data_simulasi, bins=30)
```

bins adalah rentang tiap angka, misal ada 1000 data angka 0-40, dengan bins 30, artinya rentang 0-40 dipotong/bagi 30 bagian. Jadi, tiap 1 diagram batang mempunyai lebar 1,3(range), misal sampel 1 meannya 15, maka dia masuk ke range 14-16. Kalo semakin kecil binsnya maka range/lebar diagram semakin besar, contoh sama mean 15, karena bins cuma 5 jadi rangenya masuk di 15-20


```python
# Kurva Lonceng

import numpy as np
import matplotlib.pyplot as plt

data_simulasi = np.random.normal(loc=20, scale=5, size=1000)  # simulasi 1000 data, mean=20, std=5
plt.hist(data_simulasi, bins=5)
plt.show()
```

```python
import pandas as pd

data = pd.read_csv('hasil_data_lab2.csv')

mean = data['kadar_air'].mean()
std = data['kadar_air'].std()

USL = 65   # Batas Atas Spek
LSL = 50   # Batas Bawah Spek

Cp = (USL - LSL) / (6 * std)
Cpk = min((mean - LSL) / (3 * std), (USL - mean) / (3 * std))

print(f"Mean: {mean:.2f}, Std: {std:.2f}")
print(f"Cp: {Cp:.3f}")
print(f"Cpk: {Cpk:.3f}")
```
