```python
import pandas as pd

data = pd.read_csv('data_lab.csv')

rata = data.groupby ('nama_produk')['berat_basah'].mean()

print(rata)
```

```python
 #Mencari rata rata penumpang yang selamat berdasarkan kelas

import pandas as pd

data = pd.read_csv('kapal_titanic.csv')
data = data.groupby('pclass')['survived'].mean()
print (data)
```

```python
#Menentukan rata rata usia pada setiap kelas

import pandas as pd

data = pd.read_csv('kapal_titanic.csv')

data = data.groupby('pclass')['age'].mean()

print(data)
```

```python
 #Mencari rata rata penumpang yang selamat berdasarkan jenis kelamin

import pandas as pd

data = pd.read_csv('kapal_titanic.csv')
data = data.groupby('sex')['survived'].mean()
print (data)
```

```python
#Gabungan 2 data yaitu kelas tiket dengan jenis kelamin, mencari rata rata yang selamat.

import pandas as pd

data = pd.read_csv('kapal_titanic.csv')

data = data.pivot_table(values='survived', index='pclass', columns='sex', aggfunc='mean')

print(data)
```
