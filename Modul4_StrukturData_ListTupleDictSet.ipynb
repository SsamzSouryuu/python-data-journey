```python
# Proyek Modul 4, Rekap banyak sampel sekaligus (pake gabungan dari modul 2,3, dan 4)

sampel1 = {
    "nama_produk": "Keripik Pisang",
    "berat_basah": 87.2,
    "berat_kering": 32.6,
    "brix": 18
}

sampel2 = {
    "nama_produk": "Keripik Tempe",
    "berat_basah": 98.2,
    "berat_kering": 43,
    "brix": 14
}

sampel3 = {
    "nama_produk": "Keripik Apel",
    "berat_basah": 100,
    "berat_kering": 45,
    "brix": 22
}

list_sampel = [sampel1, sampel2, sampel3]


def hitung_kadar(berat_basah, berat_kering):
  hitung_kadar = ((berat_basah - berat_kering)/berat_basah)*100
  return hitung_kadar

def klasifikasi_brix(hasil):
    if hasil >= 20:
      return "A"
    elif hasil >= 15:
      return "B"
    else:
      return "C"

for semua_sampel in list_sampel:
  kadar_air = hitung_kadar(semua_sampel["berat_basah"],semua_sampel["berat_kering"])
  print (kadar_air)
  hasil = klasifikasi_brix(semua_sampel["brix"])
  print (hasil)

  print (f"Nama Produk: {semua_sampel['nama_produk']}")
  print (f"Brix: {semua_sampel['brix']}")
  print (f"Kadar Air: {kadar_air: .2f}")
  print (f"Grade: {hasil}")
```
