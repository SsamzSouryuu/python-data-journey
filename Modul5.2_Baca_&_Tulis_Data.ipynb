```python
import csv
import pandas as pd

def hitung_kadar(berat_basah, berat_kering):
  kadar_air = ((berat_basah - berat_kering) / berat_basah) * 100
  return kadar_air

def klasifikasi_brix(hasil):
  if hasil >= 20:
    return "A"
  elif hasil >= 15:
    return "B"
  else:
    return "C"

#Menyuruh python untuk membuka file dan membuat file baru bernama "Laporan_reject.csv"
with open ('data_lab_asli.csv',mode='r') as file_sumber, open ('Laporan_reject.csv',mode='w', newline='') as file_tujuan:

  pembaca = csv.DictReader(file_sumber)

  field_kolom = ['nama_produk', 'berat_basah', 'berat_kering', 'brix', 'kadar_air', 'grade']

  penulis = csv.DictWriter(file_tujuan, fieldnames=field_kolom)

#Menyuruh python untuk menulis headernya dulu di file baru.
  penulis.writeheader()

#Menyuruh python untuk menghitung dan percabangan/menentukan hasil.
  for hasil_data in pembaca:
    kadar_air = hitung_kadar(float(hasil_data["berat_basah"]), float(hasil_data["berat_kering"]))
    hasil = klasifikasi_brix(int(hasil_data["brix"]))

    if hasil == "C":
      hasil_data["kadar_air"] = round(kadar_air, 2)
      hasil_data["grade"] = hasil

      penulis.writerow(hasil_data)


print("Laporan Reject berhasil dieksekusi tanpa error!")
```
