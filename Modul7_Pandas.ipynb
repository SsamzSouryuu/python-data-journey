```python
import pandas as pd

data = pd.read_csv('data_lab.csv')

hasil_data = data.describe()

kadar_air = data['kadar_air'] = ((data['berat_basah'] - data['berat_kering']) / data['berat_basah']) * 100

def klasifikasi_brix(brix): #Nentuin grade untuk brix
    if brix >= 20:
        return "A"
    elif brix >= 15:
        return "B"
    else:
        return "C"

data['grade'] = data['brix'].apply(klasifikasi_brix) #nyuruh nambah header baru yaitu 'grade', yang terambil dari kolom brix dan mengaplikasikan pada 'def'nentuin grade.
print(data[['brix', 'grade']]) #ngeprint brix dan grade

reject = data[data['grade'] == 'C'] #Ngeprint khusus reject/grade C doang.
print(reject)

jumlah_reject = len(reject) #Ngambil seluruh/total yang kena reject doang
jumlah_total = len(data) #Ngambil seluruh data
persen_reject = (jumlah_reject / jumlah_total) * 100

print(f"Reject: {jumlah_reject} dari {jumlah_total} sampel ({persen_reject:.2f}%)")

reject.to_csv('laporan_reject_pandas.csv', index=False) #Nyuruh pandas untuk buat file baru
```
