```python
import pandas as pd

data = pd.read_csv('data_lab.csv')

#bikin kolom baru kadar_air, ngitung rumus, dan terapin ke semua baris
data['kadar_air'] = ((data['berat_basah'] - data['berat_kering']) / data['berat_basah']) * 100

Q1 = data['kadar_air'].quantile(0.25) #quantile(0.25) itu nyari nilai di posisi 25% kalau semua data kadar air diurutin dari kecil ke besar.
#bayangin 25 sampel kadar air diurutin dari terkecil ke terbesar, berbaris rapi. Q1 itu nilai yang ada di posisi seperempat pertama barisan itu — artinya 25% data ada di bawah nilai ini.

Q3 = data['kadar_air'].quantile(0.75) #Sama konsepnya, tapi posisi 75% — 75% data ada di bawah nilai ini (atau bisa dibilang, 25% data "teratas" ada di atas nilai ini).

IQR = Q3 - Q1 #IQR (Interquartile Range) itu selisih antara Q3 dan Q1 — ngukur seberapa lebar "badan tengah" data

#pagar toleransi.
batas_bawah = Q1 - (1.5 * IQR)
batas_atas = Q3 + (1.5 * IQR)

#boolean indexing, gabungan dua kondisi pakai | (artinya "atau"). Pandas pakai simbol |, bukan kata or
#adi baris ini nyaring data yang kadar airnya kelewat rendah ATAU kelewat tinggi dari pagar toleransi — dua-duanya dianggap outlier.
outlier = data[(data['kadar_air'] < batas_bawah) | (data['kadar_air'] > batas_atas)]
print(outlier)
```
