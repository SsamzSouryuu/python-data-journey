```python
import matplotlib.pyplot as plt
import numpy as np

kadar_air = np.array([15, 20, 18, 23, 21, 42, 12, 23, 20, 21, 32, 12, 12, 15, 17])

mean = kadar_air.mean()
std = kadar_air.std()
UCL = mean + (2 * std)
LCL = mean - (2 * std)

plt.plot(kadar_air, marker = 'o')
plt.axhline(mean, color = 'green', linestyle = '-', label = 'Mean')
plt.axhline(UCL, color = 'red', linestyle = '-', label = 'UCL')
plt.axhline(LCL, color = 'orange', linestyle = '-', label = 'LCL')
plt.xlabel('Batch ke-')
plt.ylabel('Kadar Air (%)')
plt.title('Control Chart Kadar Air')
plt.legend()
plt.show()
```

```python
# Project 12
# Membuat control chart sama persis tapi dengan file
# Tentuin titik titik yang keluar dan titik berurutan (konsisten/pola run)

import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv('hasil_data_lab2.csv')

mean = data['kadar_air'].mean()
std = data['kadar_air'].std()
UCL = mean + (2 * std)
LCL = mean - (2 * std)

plt.plot(data['kadar_air'], marker='o')
plt.axhline(mean, color='green', linestyle='-', label='Mean')
plt.axhline(UCL, color='red', linestyle='--', label='UCL')
plt.axhline(LCL, color='orange', linestyle='--', label='LCL')
plt.xlabel('Sampel ke-')
plt.ylabel('Kadar Air (%)')
plt.title('Control Chart Kadar Air (Data Lab)')
plt.legend()
plt.show()
```

1. Titik yang keluar dari "pagar" adalah sampel nomer 14 yang memiliki kadar air sekitar 71 sekian dan sampel nomer 16 yang memiliki kadar air sekitar 40 sekian. dan tidak ada pola 'run' dalam hasil lab tersebut.
